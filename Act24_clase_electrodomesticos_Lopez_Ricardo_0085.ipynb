Descripción del Proyecto
El objetivo es desarrollar un sistema de gestión de dispositivos utilizando Programación Orientada a Objetos (POO) en Python, modelando la clase Electrodomestico para integrar sus características y comportamientos en una sola unidad funcional. Mediante el uso de atributos (nombre, marca, consumoEnergia, garantiaAnos) se define la identidad técnica de cada aparato, mientras que las funciones (encender(), programarTemporizador(), diagnosticarFallo()) permiten simular su operación real y validar su estado lógico. Este enfoque permite crear múltiples objetos independientes, garantizando que cada dispositivo opere de forma autónoma, organizada y eficiente.

![texto del vínculo](https://raw.githubusercontent.com/RicardoLopezCalderon128/Act24-clase-electrodomesticos-Lopez-Ricardo-0085/refs/heads/main/infografia%20electrodomesticos.png)

```python
class Electrodomestico:
    def __init__(self, nombre, marca, consumo, garantia):
        self.nombre = nombre
        self.marca = marca
        self.consumoEnergia = consumo
        self.garantiaAnos = garantia
        self.esta_encendido = False

    def encender(self):
        if not self.esta_encendido:
            self.esta_encendido = True
            return f"⚡ [SISTEMA]: El/La {self.nombre} {self.marca} se ha encendido."
        else:
            return f"⚠️ [AVISO]: El/La {self.nombre} ya estaba encendido."

    def programar_temporizador(self, minutos):
        if self.esta_encendido:
            return f"⏰ [TIMER]: Temporizador de {self.nombre} ajustado a {minutos} min."
        else:
            return f"❌ [ERROR]: No se puede programar. {self.nombre} está apagado."

    def diagnosticar_fallo(self):
        estado = "Funcionando" if self.esta_encendido else "Modo Ahorro"
        return f"🔍 [STATUS]: Diagnosticando {self.nombre} ({self.marca}). Estado: {estado}. Garantía: {self.garantiaAnos} años."

# --- Menú Interactivo para 2 Objetos ---

print("=== REGISTRO DE ELECTRODOMÉSTICOS ===")

# --- Creación del Electrodoméstico 1 ---
print("\n--- Datos para el Electrodoméstico 1 ---")
n1 = input("Tipo: "); m1 = input("Marca: "); c1 = input("Consumo: "); g1 = input("Garantía: ")
electro1 = Electrodomestico(n1, m1, c1, g1)

# --- Creación del Electrodoméstico 2 ---
print("\n--- Datos para el Electrodoméstico 2 ---")
n2 = input("Tipo: "); m2 = input("Marca: "); c2 = input("Consumo: "); g2 = input("Garantía: ")
electro2 = Electrodomestico(n2, m2, c2, g2)

# --- Pruebas de funcionamiento ---

print("\n" + "="*40)
print(f"PRUEBAS PARA {electro1.nombre.upper()}")
print(electro1.encender())
print(electro1.programar_temporizador(30))
print(electro1.diagnosticar_fallo())

print("\n" + "="*40)
print(f"PRUEBAS PARA {electro2.nombre.upper()}")
print(electro2.programar_temporizador(10)) # Intentar programar antes de encender
print(electro2.encender())
print(electro2.diagnosticar_fallo())
print("Programa realizado por Ricardo Lopez Calderon NL = 29")
print("\n================================================")
```
